Welcome to **eidcloud-linux-troubleshooter**! This Google Colab notebook demonstrates how the autonomous agent safely diagnoses and remediates common Linux server crises, including **Nginx 502 Bad Gateway**, **PHP-FPM socket starvation**, **OOM kill events**, and **storage exhaustion**.

### Safe Execution Lifecycle:
1. **READ**: Inspects `journalctl`, `systemctl`, `dmesg`, Nginx/PHP-FPM logs, network ports (`ss`), and resource metrics (`free`/`df`).
2. **DIAGNOSE**: Evaluates rules against live telemetry, computing confidence scores and isolating log evidence.
3. **PROPOSE**: Synthesizes an exact, idempotent remediation plan.
4. **CONFIRM**: Prompts for operator confirmation or executes with `--auto-remediate`.
5. **EXECUTE & VERIFY**: Applies minimal shell commands and re-runs diagnostics to verify service health.
